```python
#Inicializo los diccionarios que contabilizan el puntaje y victorias finales de cada participante
nombre_de_participantes = list(rounds[0]['scores'].keys())

puntajes_total = {clave: 0 for clave in nombre_de_participantes}
victorias_total = {clave: 0 for clave in nombre_de_participantes}
maximo = {clave: 0 for clave in nombre_de_participantes}

#Iteración entre rondas
for indice,datos in enumerate(rounds):
    print(f'Ronda {indice+1} - {datos['theme']}')
    nombre_ganador, parcial = competencia_ej10.simular_ronda(datos['scores'], nombre_de_participantes)
    
    #Adición del puntaje parcial al puntaje total y contabilización del puntaje record individual
    for indice,clave in enumerate(puntajes_total):
        puntajes_total[clave] += parcial[indice]
        if parcial[indice] > maximo[clave]:
            maximo[clave] = parcial[indice]

    #Adición de victorias
    victorias_total[nombre_ganador] += 1
    
print(puntajes_total)
print(maximo)
print(victorias_total)
```
